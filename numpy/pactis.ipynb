```python
X = [[1,2,3,2.5],[2,4,2.6,-1.9],[2.6,1.5,-2.5,0.9]]
```

```python
class Layer_Dense:
    def __init__(self,n_input,n_nuren):
        self.weight = np.random.rand(n_input,n_nuren)
        self.bises = np.zeros((1,n_nuren))
    def forword(self,inputs):
        self.output = np.dot(inputs,self.weight) + self.bises
```

```python
layer1 = Layer_Dense(4,6)
layer1.forword(X)
print(layer1.output)
```

```python
layer2 = Layer_Dense(6,5)
layer2.forword(layer1.output)
print(layer2.output)
```

```python
layer3 = Layer_Dense(5,2)
layer3.forword(layer2.output)
print(layer3.output) # this is the final output of the last layer.
```
