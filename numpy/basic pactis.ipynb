```python
# take a user input for creating a matrix , the shape, the number want to create .
row = int (input("Enter the number of rows :"))
columns = int (input("Enter the number of columns :"))
number = int(input("Enter the number that you whould like to fill with :"))
user_matrix = np.full((row,columns),number)
print(user_matrix)
```
