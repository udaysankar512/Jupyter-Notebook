```python
def join_after(date):
    if date > 2015:
        return 'New'
    elif date < 2015:
        return 'Old'
    else:
       return 'Mid'
```

```python
print("The new joiners years :",main_data['JoiningYear'].max())
print("The older joiners year :",main_data['JoiningYear'].min())
```

```python
main_data['Joning_Status'] = main_data['JoiningYear'].apply(join_after)
main_data.head()
```

```python
n = main_data.groupby('Joning_Status')['Age'].mean()
n
```

```python
m = main_data.groupby('Joning_Status')['Age'].mean().reset_index()
m
```

```python
def joiningRate(date):
    if date == 2018 or date == 2012 or date == 2016:
        return "Low"
    elif 2013 <= date <= 2015:
        return "Mid"
    else:
        return "High"
```
