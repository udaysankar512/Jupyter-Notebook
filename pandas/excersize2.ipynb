```python
pandas_data = pd.DataFrame(data)
pandas_data
```

```python
# show all the employ that are working in IT department 
IT_department =pandas_data[pandas_data['Department'] == 'IT']
print("The IT employs only :")
IT_department
```

```python
# Find the average salary of Finance department employees
finans_department = pandas_data[pandas_data['Department'] == 'Finance']
finans_department_salary = finans_department['Salary']
average_salary_Fin = finans_department_salary.mean()
print("The average salary of the finans department is :",round(average_salary_Fin,2))
```

```python
# Add a new column called Bonus = 10% of Salary
pandas_data['Bonus'] = pandas_data['Salary']/10
pandas_data
```

```python
# Sort employees by Performance in descending order
sort_by_perfomence = pandas_data.sort_values(by='Performance',ascending= False)
print("The sorted employee list based on Performance is :")
sort_by_perfomence
```

```python
# Group employees by Department and calculate the average Experience
data_by_department = pandas_data.groupby('Department')['Experience'].mean()
print("The average of the department expreance is :\n ",data_by_department)
```
