```python
# TYPE OF EXPRIENCE LEVEL HAVE IN THE DATA BASE
experience_level_data = data.value_counts(data['experience_level'])
print("The count of each experience level employs are :",experience_level_data)
```

```python
# COUNT OF THE JOB TITEL / NAME HAVE IN THE DATA DASE 
jod_title_data = data.value_counts(data['job_title'])
jod_title_data
```

```python
# count the number of industery that have data engineer job titel
industry_count_job_title = data[data['job_title'] == 'Data Engineer']['industry']
print("The industry count of data engineer post are :",industry_count_job_title.value_counts())
```

```python
# show the type of employ works and there count 

employ_type = data.value_counts(data['employment_type'])
employ_type
```

```python
# print the salary 
salary_employ = data['salary_usd'].max()
print("The salary of all employ are :")
print(salary_employ)
```

```python
#print the highest salary employ ,the details we need are - industry,salary,years_experience,interviews_to_offer,certifications_count
highest_salary = data.loc[data['salary_usd'].idxmax(),
                          ['industry','salary_usd','years_experience',
                           'interviews_to_offer','certifications_count']]
print(highest_salary)
```
